### Pivot Table
The pivot table takes simple column-wise data as input, and groups the entries into a two-dimensional table that provides a multidimensional summarization of the data.

```python
import numpy as np
import pandas as pd 
import seaborn as sns
```

```python
df=sns.load_dataset('tips')
df
```

We can calculate the following things using multiindex series or groupby etc but pivot table makes them very easy 

```python
df.groupby('sex')['total_bill'].mean()
```

```python
df.groupby(['sex','smoker'])['total_bill'].mean().unstack()
```

```python
df.pivot_table(index='sex',values='total_bill',columns='smoker')
```

```python
#by default aggfunc is mean
df.pivot_table(index='sex',values='total_bill',columns='smoker',aggfunc='max')
```

```python
df.pivot_table(index=['sex','smoker'],values='total_bill').unstack()
```

```python
df.pivot_table(index='sex',columns='smoker',values='tip')
```

```python
df
```

```python
df.pivot_table(index=['sex','smoker'],values=['total_bill','tip','size'],columns='day',aggfunc={'total_bill':'mean','tip':'max','size':'min'})
```

```python
# margins=True will also calculate total off all rows and columns 
df.pivot_table(index='sex',columns='smoker',values='total_bill',aggfunc='sum',margins=True)
```

#### Plotting graphs

```python
# plotting graphs
new_df = pd.read_csv('Datasets/expense_data.csv')
new_df
```

```python
new_df['Date']=pd.to_datetime(new_df['Date'])
new_df['Month']=new_df['Date'].dt.month_name()
new_df
```

```python
new_df.pivot_table(index='Month',columns='Category',values='INR',aggfunc='sum',fill_value=0).plot()
```

```python
new_df.pivot_table(index='Month',columns='Category',values='INR',aggfunc='sum',fill_value=0,margins=True)
```
