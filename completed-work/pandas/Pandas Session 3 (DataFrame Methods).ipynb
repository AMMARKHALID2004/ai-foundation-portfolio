##### dropna() (Series and DataFrames) --> drops null values

```python
dataframe
```

```python
students = pd.DataFrame(
    {
        'name':['nitish','ankit','rupesh',np.nan,'mrityunjay',np.nan,'rishabh',np.nan,'aditya',np.nan],
        'college':['bit','iit','vit',np.nan,np.nan,'vlsi','ssit',np.nan,np.nan,'git'],
        'branch':['eee','it','cse',np.nan,'me','ce','civ','cse','bio',np.nan],
        'cgpa':[6.66,8.25,6.41,np.nan,5.6,9.0,7.4,10,7.4,np.nan],
        'package':[4,5,6,np.nan,6,7,8,9,np.nan,np.nan]

    }
)
students
```

```python
#dropna() will drop all the rows where there is atleast one nan
students.dropna()
```

```python
#how='all' parameter ensures that only those rows where every column is nan is dropped
students.dropna(how='all')
```

```python
#this will drop only those rows where there is name null OR college is null
students.dropna(subset=['name','college'])
```

```python
#this will drop the rows where there is name AND cgpa both null
students.dropna(subset=['name','cgpa'],how='all')
```

##### apply() (Series and DataFrames) --> applies a custom logic 

```python
def sigmoid(value):
  return 1/(1+np.exp(-value))
temp.apply(sigmoid)
```

```python
points_df = pd.DataFrame(
    {
        '1st point':[(3,4),(-6,5),(0,0),(-10,1),(4,5)],
        '2nd point':[(-3,4),(0,0),(2,2),(10,10),(1,1)]
    }
)

points_df
```

```python
def euclidean(row):
  pt_A = row['1st point']
  pt_B = row['2nd point']

  return ((pt_A[0] - pt_B[0])**2 + (pt_A[1] - pt_B[1])**2)**0.5
```

```python
points_df['distance'] = points_df.apply(euclidean,axis=1)
points_df
```
