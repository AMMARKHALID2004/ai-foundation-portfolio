```python
# write a function that can return the track record of 2 teams against each other
def track_record(first_team,second_team):
    h2h=ipl[((ipl['Team1']==first_team) & (ipl["Team2"]==second_team)) | ((ipl['Team1']==second_team) & (ipl["Team2"]==first_team))]
    total_matches=h2h.shape[0]
    first_team_wins=h2h[h2h["WinningTeam"]==first_team].shape[0]
    second_team_wins=h2h[h2h["WinningTeam"]==second_team].shape[0]
    return f"Head-to-Head: {first_team} vs {second_team}\n" \
           f"Total Matches: {total_matches}\n" \
           f"➔ {first_team}: {first_team_wins} wins\n" \
           f"➔ {second_team}: {second_team_wins} wins"

print(track_record("Mumbai Indians","Rajasthan Royals"))
    
```
