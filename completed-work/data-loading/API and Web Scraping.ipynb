#### Working with APIs

```python
import pandas as pd
import requests
```

```python
import os
url = "https://www.omdbapi.com/?i=tt3896198"
params = {"apikey": os.environ["OMDB_API_KEY"]}
headers={'User-Agent':'Mozilla/5.0 (X11; Linux x86_64; rv:151.0) Gecko/20100101 Firefox/151.0',
         'Accept':'Application/json'
        }
response=requests.get(url, headers=headers, params=params, timeout=30)
```

```python
json=response.json()
pd.DataFrame(json)
```
