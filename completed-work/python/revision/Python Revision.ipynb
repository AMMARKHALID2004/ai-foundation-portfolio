## Problem 1 — Generators: chunked
What "chunked" means: splitting a long sequence into smaller, fixed-size groups. Example: the list [1,2,3,4,5,6,7] split into chunks of size 3 becomes [1,2,3], [4,5,6], [7] (the last group can be smaller if it doesn't divide evenly).

The problem: write a generator function chunked(iterable, size) that yields successive chunks of length size from iterable, as lists.

```python
from itertools import batched

def chunked(iterable,size):
    for i in batched(iterable,size):
        yield i

l=[1,2,3,4,5,6,7]

demo=chunked(l,3)

print(next(demo))
print(next(demo))
print(next(demo))

for chunk in chunked(l,3):
    print(chunk)
```

```python
L=[1,2,3,4,5,6,7]

def chunked(iterable,size):
    for i in range(0,len(iterable),size):
        yield iterable[i:i+size]

demo=chunked(L,3)

print(next(demo))
print(next(demo))
print(next(demo))
    
```

## Problem 2 — Decorators: count_calls
The problem: write a decorator count_calls that tracks how many times the function it decorates has been called, across all calls, and lets you check that count afterward.

```python
def count_calls(func):
    def wrapper(*args,**kwargs):
        result=func(*args,**kwargs)
        wrapper.calls+=1   # when the decorator is applied to a function, the code inside it runs and 
                          # when that function is called then the code inside the wrapper runs
        return result
    wrapper.calls=0
    return wrapper


@count_calls
def greet():
    return "hi"

greet()
greet()
greet()
print(greet.calls)
```

```python
def digit_sum(number):
    if number<10:
        return number
    dig=number%10
    number=number//10
    return dig + digit_sum(number)
   

digit_sum(1234)
```

## Problem 2: 
write safe_divide(a, b)

```python
def safe_divide(a,b):
    try:
        division=a/b
    except ZeroDivisionError:
        return "Cannot divide by zero"
    except TypeError:
        return "Invalid input type"
    else: # else only runs when there is no exception thrown
        print(f"Division successful: {division}")
        return division
    finally:
        print("Operation complete")

print(safe_divide(10, 2))     # works fine
print(safe_divide(10, 0))     # triggers ZeroDivisionError
print(safe_divide(10, "a"))   # triggers TypeError
```

## Problem 1 — File Handling: word frequency counter
##### Variant 1

```python
def word_count(filepath):
    counts={}
    with open(filepath) as f:
            l=f.read().lower().split()
    for i in l:
        if i in counts:
            counts[i]+=1
        else:
            counts[i]=1
       
    print(counts)
    
        
            

word_count("file.txt")
```

##### Variant 2

```python
from collections import Counter

def word_count(filepath):
    with open(filepath) as f:
        l=f.read().lower().split()
    return Counter(l)

word_count("file.txt")
```

## Problem 2 — Pathlib 

```python
from pathlib import Path

def list_text_files(folder_path):
    l2=[]
    path=Path(folder_path)
    if not path.exists():
        return l2
    for p in path.glob("*.txt"):
        l2.append(p.name)
    return l2

result_=list_text_files(Path.home()/"Python Revision")
print(result_)
```

## Problem 3 — Regex


##### Part A — extract all numbers from a string:

```python
import re

pattern=re.compile(r"\d+")

def extract_numbers(text):
    l3=[]
    matches=pattern.finditer(text)
    for match in matches:
        l3.append(match.group(0))
    return l3

extract_numbers("I have 12 apples and 5 oranges, total cost is 350 dollars")
    
```

##### Part B — validate a simple email pattern:

```python
email_pattern=re.compile(r"^[A-Za-z0-9]+\.?[A-Za-z0-9]+@[A-Za-z]+\.[A-Za-z]+$")
def is_valid_email(text):
    if email_pattern.fullmatch(text):
        return True
    else:
        return False


print(is_valid_email("ali123@gmail.com"))   # True
print(is_valid_email("not an email"))       # False
print(is_valid_email("missing@dotcom"))     # False
    
```

##### Problem 2 — OOP: build a BankAccount class

```python
class BankAccount:
    def __init__(self,owner,balance=0):
        self.owner=owner
        self._balance=balance
        
    def deposit(self,amount):
            if amount>0:
                self._balance+=amount
            else:
                raise ValueError("Amount must be positive")

    def withdraw(self,amount):
            if amount<=self._balance:
                self._balance-=amount
            else:
                 raise ValueError("Insufficient funds")
                
    
    def get_balance(self):
        return self._balance
    
    def __str__(self):
        return f"Account owner:{self.owner}, Balance:{self._balance}"

acc = BankAccount("Ali", 100)
acc.deposit(50)
acc.withdraw(30)
print(acc.get_balance())

print(acc)

acc.withdraw(1000)
```
