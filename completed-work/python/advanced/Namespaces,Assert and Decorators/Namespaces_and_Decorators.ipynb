## Namespaces

A namespace is a space that holds names(identifiers).Programmatically speaking, namespaces are dictionary of identifiers(keys) and their objects(values)

There are 4 types of namespaces

    * Builtin Namespace
    * Global Namespace
    * Enclosing Namespace
    * Local Namespace


## Scope and LEGB Rule

A scope is a textual region of a Python program where a namespace is directly accessible.

The interpreter searches for a name from the inside out, looking in the local, enclosing, global, and finally the built-in scope. If the interpreter doesn’t find the name in any of these locations, then Python raises a NameError exception.

```python
# local and global
# global var
a = 2

def temp():
  # local var
  b = 3
  print(b)

temp()
print(a)
```

```python
# local and global -> same name
a = 2

def temp():
  # local var
  a = 3
  print(a)

temp()
print(a)
```

```python
# local and global -> local does not have but global has
#it will follow LEGB rule
a = 2

def temp():
  # local var
  print(a)

temp()
print(a)
```

```python
# we cannot write on a global variable inside a local scope but we can only read it

a=9

def fun():
    a+=1
    print(a)
fun()
```

```python
# global keyword provides the access to write on a global variable inside a local scope

a=9

def fun():
    global a
    a+=1
    print(a)
fun()

# we can also create a global variable inside a local scope

def function():
    global b
    b=1
    print(b)
# first we would have to call that function and only then we can access the global variable created inside that function
function()
print(b)
```

## Decorators

A decorator in python is a function that receives another function as input and adds some functionality(decoration) to and it and returns it.

This can happen only because python functions are 1st class citizens.

There are 2 types of decorators available in python:

    * Built in decorators like @staticmethod, @classmethod, @abstractmethod and @property etc
    * User defined decorators that we programmers can create according to our needs



In Python, functions are called first-class citizens (or first-class objects) because they can be treated exactly like any other data type, such as strings, integers, or lists. This means functions do not have restrictions or "second-class" status; they share the same rights, capabilities, and flexibilities as regular variables.According to programming language theory, an entity earns first-class citizen status if it satisfies four core criteria:

You can reference a function by assigning it to a variable without executing it (omitting the parentheses). The new variable becomes an alias and points to the exact same memory address.

```python
def greet(name):
    return f"Hello, {name}!"

# Assign the function to a variable
say_hello = greet

# Call the function using the new variable
print(say_hello("Alice"))  # Output: Hello, Alice!
```

You can pass a function directly into another function as a parameter. This mechanism powers "higher-order functions" and is heavily utilized for callbacks, sorting criteria, and event-driven frameworks.

```python
def shout(text):
    return text.upper()

# This function accepts another function as an argument
def format_message(func, message):
    return func(message)

print(format_message(shout, "welcome"))  # Output: WELCOME
```

A function can dynamically generate or contain a nested "inner function" and then return that inner function object as an output value. This behavior forms the absolute foundation for advanced structural features like closures and decorators

```python
def multiplier(factor):
    def multiply(number):
        return number * factor
    return multiply  # Returns the inner function object

double = multiplier(2)
print(double(5))  # Output: 10
```

Functions can be safely stored inside lists, dictionaries, tuples, or sets. This capability lets developers map operations dynamically, which is frequently used to simulate switch-case architectures.

```python
def add(a, b): return a + b
def subtract(a, b): return a - b

# Store the functions inside a dictionary
operations = {
    "sum": add,
    "diff": subtract
}

# Retrieve and execute
print(operations["sum"](10, 5))  # Output: 15
```

```python
# simple example

def my_decorator(func):
  def wrapper():
    print('***********************')
    func()
    print('***********************')
  return wrapper

def hello():
  print('hello')

def display():
  print('hello nitish')

a = my_decorator(hello)
a()

b = my_decorator(display)
b()
```

```python
def decorator1(func):
    def wrapper():
        print("************")
        func()
        print("****************")
    return wrapper
#decorators can be simplified by using this syntax
@decorator1
def hi():
    print("hi")
hi()



```

```python
import time

def timer(func):
    def wrapper(*args):
        start=time.time()
        func(*args)
        print("Time taken by the ",func.__name__," is ",time.time()-start)
    return wrapper

@timer
def greet():
    print("greetings")
    time.sleep(2)
    
@timer
def max(a,b):
    print("max")
    time.sleep(3)
    if a>b:
        return a
    else:
        return b
        
max(9,10)

greet()
```

```python
def sanity_check(data_type):
  def outer_wrapper(func):
    def inner_wrapper(*args):
      if type(*args) == data_type:
        func(*args)
      else:
        raise TypeError('Ye datatype nai chalega')
    return inner_wrapper
  return outer_wrapper

@sanity_check(int)
def square(num):
  print(num**2)

@sanity_check(str)
def greet(name):
  print('hello',name)

square(2)
```
