sum/min/max/prod

```python
arr1=[[23,45,6,2],[12,24,45,11]]
np.sum(arr1)
```

```python
print(np.max(arr1))
```

```python
#axis=1-->row and axis=0-->column
np.min(arr1,axis=1)
```

```python
np.prod(arr1,axis=0)
```

Statistical Operations

```python
np.mean(arr1)
```

```python
np.std(arr1,axis=1)
```

```python
np.var(arr1,axis=1)
```

Trigonometric Functions (not very useful in our journey for data science)

```python
np.cos(arr1)
```

Dot product (can be performed only when the no. of columns of first matrix = no. of rows of second matrix 
The final matrix will be of order (no. of rows of first matrix x no. of rows of second matrix)

```python
arr2=np.array([[1,3],[12,55],[2,34],[234,35]])
```

```python
np.dot(arr1,arr2)
```

Log and Exponents

```python
np.log(arr2)
```

```python
#Returns e^x
np.exp(arr2)
```

Floor,Ceil,Round

```python
np.floor(arr2/2)
```

```python
np.ceil(arr2/2)
```
