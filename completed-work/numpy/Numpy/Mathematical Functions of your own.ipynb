## Mathematical functions of your own

#### We can make any mathematical function using Numpy which can we performed on any number of elements


```python
import numpy as np
arr=np.random.randint(1,50,20).reshape(4,5)
print(arr)
```

### Sigmoid function

#### Sigmoid function is used as an activation function in neural networks and is also the core 
#### part of logistic regression algorithm

```python
def sigmoid(array):
    return 1/(1+np.exp(-(array)))
```

```python
sigmoid(arr)
```

### Mean squared error

#### Mean squared error is used as a loss function in linear regression algorithm of Machine Learning

```python
#Actual output
actual=np.arange(20).reshape(4,5)
#Predicted output
predicted=np.arange(20,40).reshape(4,5)

print(actual)
print(predicted)

```

```python
def mse(actual,predicted):
    return np.mean((actual-predicted)**2)
```

```python
mse(actual,predicted)
```

### Binary cross entropy

#### Binary cross entropy is used as a loss function in binary classification

```python
actual_binary_label=np.random.randint(0,2,25).reshape(5,5)
predicted_probability=np.random.random(size=(5,5))
print(actual_binary_label)
print(predicted_probability)
```

```python
def bce(actual,predicted):
    return -(np.mean(actual*np.log(predicted)+np.multiply(np.subtract(1,actual),np.log(np.subtract(1,predicted)))))
```

```python
bce(actual_binary_label,predicted_probability)
```
