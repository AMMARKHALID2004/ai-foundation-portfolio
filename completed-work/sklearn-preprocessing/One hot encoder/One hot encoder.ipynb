### Train Test Split

```python
from sklearn.model_selection import train_test_split

x_train,x_test,y_train,y_test=train_test_split(df.iloc[:,0:4],df['selling_price'])
```

### OneHotEncoder using ScikitLearn

```python
from sklearn.preprocessing import OneHotEncoder

ohe=OneHotEncoder(drop='first') # drop=first for avoiding multicollinearity

ohe.fit(x_train[['fuel','owner']])

x_train_new=ohe.transform(x_train[['fuel','owner']]).toarray() # toarray() is called because sparse=True by default and we dont get a numpy array as output
x_test_new=ohe.transform(x_test[['fuel','owner']]).toarray()
```
