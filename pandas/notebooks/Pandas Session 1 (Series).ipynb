## What is Pandas

Pandas is a fast, powerful, flexible and easy to use open source data analysis and manipulation tool, built on top of the Python programming language.

https://pandas.pydata.org/about/index.html

#### Pandas Series

A Pandas Series is like a column in a table. It is a 1-D array holding data of any type.


#### Importing Pandas
We should import numpy before importing pandas a a conventional practice because pandas is built on top of Numpy

```python
import numpy as np
import pandas as pd
```

### Creating Series
##### Using lists

```python
#Strings
country=['Pakistan','USA','Australia','India','Palestine']
string_series=pd.Series(country)
string_series
# about 95% of time String is called object datatype
```

```python
#integers
marks=[100,100,34,55,74,100,95]
match_scores=pd.Series(marks)
match_scores
```

```python
#Custom index
scores=[101,43,63,73,85,109]
innings=[123,525,325,144,14,521]
match_scores=pd.Series(scores,index=innings,name="Match Scores")
```

##### Using dictionary

```python
dictionary={"Ammar":100,
            "Ahmad":91,
            "Amna":81,
            "Ayesha":80,
            "Ali":59           
           }
series=pd.Series(dictionary,name="Marks")
series
```

### Some important series methods

```python
#astype --> converts the datatype of series
import sys
print(sys.getsizeof(virat_kohli))

sys.getsizeof(virat_kohli.astype("int8"))
```

```python
#between --> checks the occurences between two certain numbers (both number are included) 
virat_kohli[virat_kohli.between(50,99)].size
```

```python
#clip --> clips the values between two numbers just like Numpy
virat_kohli.clip(50,100)
```

```python
#drop_duplicates() --> drop the duplicate values from the Series
print(subs)
#keep='first' is by default but we can also keep last and it will remove all elements before that last element
subs.drop_duplicates(keep='last')
```

```python
#duplicated()--> checks the duplicated values 
#sum() will add all True values as True=1
subs.duplicated().sum()
```

```python
#isnull()--> checks the null values 

subs.isnull().sum()
```

```python
#dropna() --> drops the null values 
temp=pd.Series([1,3,5,26,2,415,1,np.nan,12,np.nan,412,np.nan,np.nan])
print(temp)
temp.dropna()
```

```python
#fillna()--> fills the missing values

temp.fillna(1)
```

```python
#isin() checks if a certain value exists in the Series or not
#without isin()
print(virat_kohli[(virat_kohli==49) | (virat_kohli==99) | (virat_kohli==59) | (virat_kohli==9)].size)
#with isin()
virat_kohli[virat_kohli.isin([49,99,59,9])].size
```

```python
#apply() allows us to apply custom logic to our Series
virat_kohli.apply(lambda x:x**2)
```

```python
movies.apply(lambda x:'Good Movie' if x=='Shah Rukh Khan' else 'Okay Movie')
```

```python
#copy() gives the copy of the data instead of original data
new=virat_kohli.head()
new[1]=11
print(new)
print(virat_kohli) #this has changed our original data too as head() and tail() returns a view of the data instead of its copy

new_2=virat_kohli.head().copy()
new[2]=1
print(new_2)
print(virat_kohli)
```
