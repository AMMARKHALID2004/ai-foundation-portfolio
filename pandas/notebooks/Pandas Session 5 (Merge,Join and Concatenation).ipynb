### Concatenation

```python
import numpy as np
import pandas as pd
```

```python
students=pd.read_csv("~/Machine Learning/Pandas/Datasets/students.csv")
courses=pd.read_csv("~/Machine Learning/Pandas/Datasets/courses.csv")
nov=pd.read_csv("~/Machine Learning/Pandas/Datasets/reg-month1.csv")
dec=pd.read_csv("~/Machine Learning/Pandas/Datasets/reg-month2.csv")
ipl=pd.read_csv("~/Machine Learning/Pandas/Datasets/matches.csv")
deliveries=pd.read_csv("~/Machine Learning/Pandas/Datasets/deliveries.csv")
```

##### concat() --> used for concatenating two tables

```python
nov
```

```python
dec
```

```python
# ignore_index=True will make sure that your original indices are not repeated but rather new indices are made
pd.concat([nov,dec],ignore_index=True)
```

```python
# if we want to make multiple indices we can use keys parameter
pd.concat([nov,dec],keys=['Nov','Dec'])
```

```python
#concatenating horizontally 
pd.concat([nov,dec],axis=1)
```

### Merge using Joins

```python
regs=pd.concat([nov,dec],ignore_index=True)
regs
```

```python
#joins (inner,outer,left,right,self) --> same like SQL
students.merge(regs,how='inner',on='student_id')
```

##### 1. Find total revenue generated

```python
courses.merge(regs,on='course_id')['price'].sum()
```

##### 2. Find month by month revenue

```python
both_months=pd.concat([nov,dec],keys=['Nov','Dec']).reset_index()
both_months.merge(courses,on='course_id').groupby('level_0')['price'].sum()
```

##### 3. Print the registration table


```python
students.merge(regs,on='student_id').merge(courses,on='course_id')[['name','course_name','price']]
```

##### 4. Plot bar chart for revenue/course

```python
regs.merge(courses,on='course_id').groupby('course_name')['price'].sum().plot(kind='bar')
```

##### 5. Find students who enrolled in both the months

```python
nov_students=nov.merge(students,on='student_id')['name']
nov_students
```

```python
dec_students=dec.merge(students,on='student_id')['name']
dec_students
np.intersect1d(nov_students,dec_students)
```

##### 6. find course that got no enrollment

```python
np.setdiff1d(courses['course_name'],regs.merge(students,on='student_id').merge(courses,on='course_id')['course_name'])
```

##### 7. Find students who did not enroll into any courses

```python
np.setdiff1d(students['name'],students.merge(regs,on='student_id')['name'])
```

##### 8. Print student name -> partner name for all enrolled students

```python
students.merge(students,left_on='student_id',right_on='partner')[['name_x','name_y']]
```

##### 9. Find top 3 students who did most number enrollments

```python
regs.merge(students,on='student_id').groupby('name').count()['student_id'].sort_values(ascending=False).head(6)
```

##### 10. Find top 3 students who spent most amount of money on courses

```python
students.merge(regs,on='student_id').merge(courses,on='course_id').groupby('name')['price'].sum().sort_values(ascending=False).head(3).reset_index()
```

### IPL Problems

##### Find top 3 studiums with highest sixes/match ratio

```python
pd.set_option('display.max_column',None)
```

```python
full_data=ipl.merge(deliveries,left_on='id',right_on='match_id')

total_sixes=full_data[full_data['batsman_runs']==6].groupby('venue')['batsman'].count()
total_matches_at_a_venue=ipl['venue'].value_counts()
(total_sixes/total_matches_at_a_venue).sort_values(ascending=False).head(3)
```

##### Find orange cap holder of all the seasons

```python
ipl.merge(deliveries,left_on='id',right_on='match_id').groupby(['season','batsman'])['batsman_runs'].sum().sort_values(ascending=False).reset_index().groupby('season').first()
```
