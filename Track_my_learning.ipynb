# My learning tracker

I select the **FYP AI (Anaconda)** kernel. I fill in my topic and note, then run the next two cells. The tracker creates my record and updates the progress log. It leaves unknown study dates as Unknown.

Module choices: `refresh`, `pipeline`, `course1`, `course2`, `unsupervised`, `pytorch`, `videos`, `deep-learning`, `evaluation`.


```python
# I fill in these values. I leave the date blank if I do not know it.
topic = ""
module = "pipeline"
note = ""  # What I learned, in my own words.
study_date = ""  # YYYY-MM-DD, or blank.
notebook = ""  # Optional path from the repository root.
```

```python
from pathlib import Path
import subprocess
import sys

root = next((p for p in [Path.cwd(), *Path.cwd().parents]
             if (p / "scripts/track_learning.py").is_file()), None)
if root is None:
    raise RuntimeError("I must open this notebook inside my portfolio folder.")
if not topic.strip() or not note.strip():
    raise ValueError("I must enter a topic and a learning note first.")
command = [sys.executable, str(root / "scripts/track_learning.py"),
           "add", topic, "--module", module, "--note", note]
if study_date:
    command += ["--date", study_date]
if notebook:
    command += ["--notebook", notebook]
subprocess.run(command, cwd=root, check=True)
```
