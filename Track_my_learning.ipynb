# My optional local learning form

My main tracker is [the GitHub checklist](https://github.com/AMMARKHALID2004/ai-foundation-portfolio/issues/1). GitHub Actions updates it when I tick a topic. This local form is optional and does not run on a schedule.

I select **FYP AI (Anaconda)**. I can fill in a topic, module, and note, then run the first two code cells to create a local record.

Module keys: `numpy`, `pandas`, `matplotlib`, `seaborn`, `ml-foundations`, `ml-data`, `ml-cleaning`, `ml-eda`, `ml-preprocessing`, `ml-pca`, `ml-regression`, `ml-classification`, `ml-trees`, `ml-ensembles`, `ml-clustering`, `ml-model-selection`, `ml-projects`.


```python
# I fill in these values. I leave the date blank if I do not know it.
topic = ""
module = "ml-preprocessing"
note = ""  # What I learned, in my own words.
study_date = ""  # YYYY-MM-DD, or blank.
notebook = ""  # Optional path from the repository root.
```

```python
from pathlib import Path
import subprocess
import sys

root = next((p for p in [Path.cwd(), *Path.cwd().parents]
             if (p / "scripts/track_learning.py").is_file()), None)
if root is None:
    raise RuntimeError("I must open this notebook inside my portfolio folder.")
if not topic.strip() or not note.strip():
    raise ValueError("I must enter a topic and a learning note first.")
command = [sys.executable, str(root / "scripts/track_learning.py"),
           "add", topic, "--module", module, "--note", note]
if study_date:
    command += ["--date", study_date]
if notebook:
    command += ["--notebook", notebook]
subprocess.run(command, cwd=root, check=True)
```
