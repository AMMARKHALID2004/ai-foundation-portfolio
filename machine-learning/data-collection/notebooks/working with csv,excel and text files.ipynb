### Working with csv files

```python
import pandas as pd
import requests
from io import StringIO
```

#### Loading csv file from the local system

```python
df=pd.read_csv('aug_train.csv')
df
```

#### Loading from the URL

```python
url = "https://raw.githubusercontent.com/cs109/2014_data/master/countries.csv"
df = pd.read_csv(url)
df
```

```python
#sometimes servers block requests, then we have to mimic a browser
url = "https://raw.githubusercontent.com/cs109/2014_data/master/countries.csv"
header={"User-Agent":"Mozilla/5.0"}
req=requests.get(url,headers=header)
data=StringIO(req.text)
df=pd.read_csv(data)
df
```

#### Sep parameter
sometimes the valus are not comma separated, so we have to tell the separator to Pandas so that our data is in the required form

```python
pd.read_csv('movie_titles_metadata.tsv')
```

```python
pd.read_csv('movie_titles_metadata.tsv',sep='\t')
```

#### names 
sometimes our data has no names for the columns, in this case we can pass our own names

```python
pd.read_csv('movie_titles_metadata.tsv',sep='\t',names=['id','name','year of release','ratings','votes','genre'])
```

#### index_col
we can make our custom index

```python
pd.read_csv('aug_train.csv',index_col='enrollee_id')
```

#### header parameter
sometimes our column names become the row, so we have to pass the row number which is actually the column to the header parameter 

```python
pd.read_csv('test.csv',header=1)
```

#### use_cols
we can tell the pandas which columns to import

```python
pd.read_csv('aug_train.csv',usecols=['gender','city'])
```

#### skiprows
we can skip the specific rows before importing

```python
pd.read_csv('aug_train.csv',skiprows=[0,1,2,3,10,16])
```

```python
pd.read_csv('aug_train.csv')
```

#### nrows 
we can tell the number of rows to be imported

```python
pd.read_csv('aug_train.csv',nrows=10)
```

#### encoding
sometimes the encoding of our data is not utf-8, so we can tell which is the encoding of the file

```python
pd.read_csv('zomato.csv')
```

```python
# try to pass latin-1, most probably it will work
pd.read_csv('zomato.csv',encoding='latin-1')
```

```python
#checking the encoding of the file
import charset_normalizer
with open('zomato.csv','rb') as f:
    print(charset_normalizer.detect(f.read(10000))) #loading only 10000 bytes in memory for detecting is enough
```

#### on_bad_lines
sometimes our number of elements in a row do not match the number of columns and it will throw
and error.We can fix this behaviour by making on_bad_lines='skip'

```python
pd.read_csv('aug_train.csv',on_bad_lines='skip')
```

#### dtype
we can change the dtype of specific column/s 

```python
pd.read_csv('aug_train.csv').info()
```

```python
pd.read_csv('aug_train.csv',dtype={"target":'int8'}).info()
```

#### handling dates

```python
pd.read_csv('IPL Matches 2008-2020.csv',parse_dates=['date']).info()
```

#### converters
we can convert specific values of a row into some other value

```python
pd.read_csv('IPL Matches 2008-2020.csv',converters={"team1":rename})
```

```python
def rename(name):
    if name=='Royal Challengers Bangalore':
        return 'RCB'
    else:
        return name
```

#### na_values
we can specify how the null values will look like


```python
pd.read_csv('aug_train.csv',na_values=['Nan','-'])
```

#### chunksize
we can load larger datasets in the memory in chunks so that our ram does not crash

```python
chunks=pd.read_csv('aug_train.csv',chunksize=5000)

for chunk in chunks:
    print(chunk.shape)
```
