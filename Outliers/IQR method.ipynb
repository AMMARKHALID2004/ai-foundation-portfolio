### IQR method
IQR method to handle the outliers is used only with skewed data

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("/home/ammar/AI Track/Machine-Learning/Outliers/Datasets/placement.csv")
```

```python
df.head()
```

### Checking skewed data

```python
fig, axes = plt.subplots(ncols=2)
sns.kdeplot(df['cgpa'], ax=axes[0])
sns.kdeplot(df['placement_exam_marks'], ax=axes[1])
```

```python
sns.boxplot(df['placement_exam_marks'])
```

```python
percentile25 = df['placement_exam_marks'].quantile(0.25)
percentile75 = df['placement_exam_marks'].quantile(0.75)
```

```python
iqr = percentile75 - percentile25
iqr
```

```python
print(percentile25)
print(percentile75)
```

```python
upper_limit = percentile75 + 1.5*(iqr)
lower_limit = percentile25 - 1.5*(iqr)
```

```python
df[df['placement_exam_marks']>upper_limit]
```

```python
df[df['placement_exam_marks']<lower_limit]
```
