* syntax wise datetime is very convenient
* But the performance takes a hit while working with huge data. List vs Numpy Array
* The weaknesses of Python's datetime format inspired the NumPy team to add a set of native time series data type to NumPy.
* The datetime64 dtype encodes dates as 64-bit integers, and thus allows arrays of dates to be represented very compactly.

```python
import numpy as np
date=np.array('2026-07-04',dtype=np.datetime64)
date
```

```python
date +np.arange(12)
```

- Because of the uniform type in NumPy datetime64 arrays, this type of operation can be accomplished much more quickly than if we were working directly with Python's datetime objects, especially as arrays get large 

- Pandas Timestamp object combines the ease-of-use of python datetime with the efficient storage and vectorized interface of numpy.datetime64

- From a group of these Timestamp objects, Pandas can construct a DatetimeIndex that can be used to index data in a Series or DataFrame

### DatetimeIndex Object

A collection of pandas timestamp

```python
#from strings
pd.DatetimeIndex(['2026/09/11','2025/09/10'])
```

```python
#using Python Date Time
pd.DatetimeIndex([dt.datetime(2026,10,10),dt.datetime(2025,11,20)])
```

```python
#using pd.Timestamp
dt_index=pd.DatetimeIndex([pd.Timestamp('2026-9-21'),pd.Timestamp('2026-9-22'),pd.Timestamp('2026-9-24')])
dt_index
```

```python
#date time index in a Series
pd.Series([1,3,5],index=dt_index)
```

### date_range() --> generates timestamps in a given range

```python
pd.date_range(start='2026-09-2',end='2026-09-06')
```

```python
# freq=2D will generate alternate timestamps and 3D will skip two just like python list
pd.date_range(start='2026-09-2',end='2026-09-06',freq='2D')
```

```python
# freq=B will generate only Business days
pd.date_range(start='2026-09-2',end='2026-09-06',freq='B')
```

```python
#freq=h will generate hourly timestamps
pd.date_range(start='2026-09-2',end='2026-09-06',freq='h')
```

```python
pd.date_range(start='2026-09-2',end='2026-09-06',freq='10h')
```

```python
#periods will specify the number of months,hours,etc
pd.date_range(start='2026-01-01',periods=25)
```

```python
# W -> one week per day (by default SUN)
pd.date_range(start='2026-09-2',end='2026-09-06',freq='W')
```

```python
pd.date_range(start='2026-09-2',end='2026-09-06',freq='W-FRI')
```

```python
# ME -> Month end
pd.date_range(start='2023/1/5',end='2023/2/28',freq='ME')
```

```python
# MS -> Month start
pd.date_range(start='2023/1/5',end='2023/2/28',freq='MS')
```

```python
# YE -> Year end
pd.date_range(start='2023/1/5',end='2030/2/28',freq='YE')
```

### to_datetime function

converts an existing objects to pandas timestamp/datetimeindex object

```python
df=pd.read_csv("~/Machine Learning/Pandas/Datasets/expense_data.csv")
df.head()
```

```python
df['Date']=pd.to_datetime(df['Date'],errors='coerce')
df.info()
```

```python
#after converting string to datetime we can apply many operations on it using the dt accessor
#errors=coerce will prevent the program from crashing if the date is Nan or some other invalid format
df['Day']=df['Date'].dt.day_name()
df.head()
```

```python
df.groupby('Day')['INR'].sum().plot(kind='bar')
```

```python
df.groupby('Account')['INR'].sum().plot(kind='bar')
```
