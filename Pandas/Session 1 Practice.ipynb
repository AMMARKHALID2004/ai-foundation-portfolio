### Problem 1 — Creating Series, three ways

```python
import numpy as np
import pandas as pd
```

```python
#using list
names=["Bob","Alice","Undertaker","John Cena"]
index=[1,2,3,4]
series_1=pd.Series(names,index=index)
series_1
```

```python
#using dictionary (automatic indexing)
marks={"Ammar":100,
      "Ahmad":90,
      "Undertaker":80}
series_2=pd.Series(marks)
series_2
```

```python
# if we have only single number as data , Pandas will repeat that data with indices

series_3=pd.Series(5,index=index)
series_3
```

### Problem 2 — Indexing: label-based vs position-based

```python
series_3 = pd.Series([100, 200, 300, 400], index=["w", "x", "y", "z"])
print(series_3.loc['y'])
print(series_3.iloc[2])
print(series_3.loc['x':'z'])
print(series_3.iloc[1:3])
```

### Problem 3 — Vectorized operations (no loops)

```python
prices = pd.Series([100, 250, 75, 300, 50], index=["a","b","c","d","e"])
print(0.9*prices)
```

```python
prices[prices>100]
```

```python
print(prices.mean())
print(prices.max())
prices.min()
```

### Problem 4 — Handling missing values in a Series

```python
data = pd.Series([10, None, 30, None, 50])
data[data.isna()].size
```
