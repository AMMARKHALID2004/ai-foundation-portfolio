##### String functions (str attribute is only for Series)

```python
df['Name'].str.capitalize()
```

```python
df['Name'].str.title()
```

##### split() anf get()

```python
# we have to add three new columns for Surname,Title and First and Middle name
df['Surname']=df['Name'].str.split(',').str.get(0) #get (0) will fetch the first item from the splitted list
df
```

```python
# n tells the the number of splits and expand=True converts the Series into a DataFrame
df[['Title','First and Middle Name']]=df['Name'].str.split(',').str.get(1).str.strip().str.split(" ",n=1,expand=True) 
df
```

##### replace() function 

```python
df['Title'].value_counts()
```

```python
df['Title']=df['Title'].str.replace('Mlle.','Miss.')
df['Title']=df['Title'].str.replace('Ms.','Miss.')
```

```python
df['Title'].value_counts()
```

```python
#na =False is a good practice that will stop the code from crashing na=False,True will handle the missing values 
df[df['Surname'].str.lower().str.startswith("a",na=False)]
```

##### regex and other searching

```python
# applying regex
# contains
# search john -> both case
df[df['firstname'].str.contains('john',case=False)]
# find lastnames with start and end char vowel
df[df['lastname'].str.contains('^[^aeiouAEIOU].+[^aeiouAEIOU]$')]
```
