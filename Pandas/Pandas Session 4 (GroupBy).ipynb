### GroupBy() --> makes a new table of every group on the basis of a categorical column

```python
import numpy as np
import pandas as pd
```

```python
movies=pd.read_csv("~/Machine Learning/Pandas/Datasets/imdb-top-1000.csv")
movies
```

```python
genre=movies.groupby("Genre")
genre
```

##### Applying builtin aggregation fuctions on groupby objects

```python
genre.std(numeric_only=True)
```

```python
genre.mean(numeric_only=True).head(3)
```

##### find the top 3 genres by total earning

```python
genre['Gross'].sum().sort_values(ascending=False).head(3)
```

#####  find the genre with highest avg IMDB rating

```python
genre['IMDB_Rating'].mean().sort_values(ascending=False).head(1)
```

##### find director with most popularity

```python
movies.groupby('Director')['No_of_Votes'].sum().sort_values(ascending=False).head(1)
```

```python
movies.groupby('Star1')['Series_Title'].count().sort_values(ascending=False)
```

##### agg() --> applies different aggregate functions on a groupby object at once

```python
genre.agg({"Runtime":'mean',
          "Gross":"sum"})
```

```python
#we can pass a list to apply many aggregate functions to the same columns
genre.agg({"Runtime":['mean','max'],"Gross":['sum','mean']})
```

##### apply()
this follow split,apply and combine approach

```python
genre['IMDB_Rating'].apply(max)
```

```python
# find number of movies starting with A for each group
genre.apply(lambda x: (x['Series_Title'].str.startswith('A')).sum()) # str is applicable only on Series not DataFrames
```

```python
# find ranking of each movie in the group according to IMDB score
def ranking(group):
    group['Rank']=group["IMDB_Rating"].rank(ascending=False)
    return group
genre.apply(ranking)
```

```python
# find normalized IMDB rating group wise
genre.apply(lambda x:(x["IMDB_Rating"]-x["IMDB_Rating"].min())/(x["IMDB_Rating"].max()-x["IMDB_Rating"].min()))
```

##### groupby() on multiple cols

```python
duo=movies.groupby(['Star1','Director'])
duo.get_group(('Aamir Khan', 'Nitesh Tiwari'))
print(duo.size())
duo.get_group(('Christian Bale','Christopher Nolan'))
```

```python
list(duo.groups)
```

```python
# find the most earning actor->director combo
duo['Gross'].sum().sort_values(ascending=False).head(1)
```

```python
# find the best(in-terms of metascore(avg)) actor->genre combo
movies.groupby(['Star1','Genre'])['Metascore'].mean().sort_values(ascending=False).head(10)
```

```python
# agg on multiple groupby
duo.agg({"Gross":["min",'max','mean']})
```

```python
psl=pd.read_csv("~/Machine Learning/Pandas/Datasets/PSL.csv")
psl
```

```python
psl.head()
```

```python
pd.set_option("display.max_columns",None)
```

```python
# find the top 10 batsman in terms of runs
psl.groupby('batter')['batsman_runs'].sum().sort_values(ascending=False).head(10)
```

```python
# find the batsman with max no of sixes
six=psl[psl['batsman_runs']==6]
six.groupby('batter')['match_id'].count().sort_values(ascending=False).head(1)
```

```python
# find batsman with most number of 4's and 6's in last 5 overs
death_overs=psl[psl['over']>15]
death_batters=death_overs[(death_overs['batsman_runs']==6) | (death_overs['batsman_runs']==4)]
death_batters.groupby('batter')['match_id'].count().sort_values(ascending=False)
```

```python
# find Babar Azam's record against all teams
bobzy=psl[psl['batter']=='Babar Azam']
bobzy.groupby('bowling_team')['batsman_runs'].count().sort_values(ascending=False)
```

```python
# Create a function that can return the highest score of any batsman
def highest_score(batsman):
    bt=psl[psl['batter']==batsman]
    return bt.groupby('match_id')['batsman_runs'].sum().sort_values(ascending=False).head(1).values[0]

highest_score('Imad Wasim')
```
