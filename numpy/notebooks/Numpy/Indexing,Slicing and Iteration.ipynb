```python
import numpy as np
arr=np.arange(24).reshape(6,4)
arr
```

```python
#In 2-d arrays index will return whole row of the matrix
arr[0]
```

```python
#for accessing third column of the matrix we will print all the rows and from each row we will print 3rd element 
arr[:,2]
```

```python
arr[:,2:4]
```

```python
#Accessing 9,10,13,14
arr[2:4,1:3]
```

For printing elements in an nd array , we  use  a fucntion np.nditer() which will print all elements

```python
#simple for loop will print whole rows of a matrix
for i in arr:
    print(i)

#Numpy method
for i in np.nditer(arr):
    print(i)
```

```python
#we want 1,3,9,11
arr_test=np.arange(0,12).reshape(3,4)
print(arr_test)
arr_test[::2,1::2]
```

For 3-d Matrices

```python
arr2=np.array([[[1,2,3,4],[5,6,7,8]],[[10,11,12,13],[14,15,16,17]]])
print(arr2)
```

```python
arr2[1]
```

```python
#Printing 5,6 from this matrix
arr2[0][1,0:2]
```

```python
#Iteration method
for i in arr2:
    print(i)
```

```python
for i in np.nditer(arr2):
    print(i)
```

```python
arr3=np.arange(27).reshape(3,3,3)
print(arr3)
```

```python
# we want 0,2,18,20
arr3[0::2,0,0::2]
```

```python
arr4=np.arange(16).reshape(2,2,2,2)
print(arr4)
```

```python
#We want 8,9,12,13
arr4[1,0:,0,0:]
```
