Numpy arrays are better than Python native lists in terms of
i)Time
ii)Memory
iii)Convenience


Testing Memory consumption


```python
import numpy as np
import time
import sys
#here I put 1 in getsizeof function because 1 is an element in the list and we have to get size of one element * length of list to get 
#total memory consumed 
list=range(1000)
print(sys.getsizeof(1)*len(list))
```

```python
arr=np.arange(1000)
print(arr.itemsize*arr.size)
#same function
print(arr.nbytes)
```

Testing time consumption

```python
list2=range(1000000)
list3=range(1000000,2000000)
start_time=time.time()
list4=[i+j for i,j in zip(list3,list3)]
print(time.time()-start_time)
```

```python
arr2=np.arange(1000000)
arr3=np.arange(1000000,2000000)
starttime=time.time()
c=arr2+arr3
print(time.time()-starttime)
```
