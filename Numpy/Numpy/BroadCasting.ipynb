# Broadcasting

The term broadcasting describes how numpy treats arrays with different shapes during the arithmetic operations.
The smaller array is broadcasted upon the larger array to have compatible types

## Broadcasting rules

#### 1.Make the two arrays have the same number of dimensions.
+ If the number of dimensions of the two arrays are different add new dimensions of size 1 to the head of the array with smaller dimensions 

#### 2.Make each dimesion of the two arrays the same size.
+ If the sizes of the each dimension of two arrays do not match,dimensions with size 1 are stretched to the size of the corresponding array's dimension
+ If there is a dimesion whose size not 1 in either of the arrays then broadcasting cannot happen and an error is raised

#### These two arrays cannot be broadcasted because if we add 1 to head of 2nd array and then extend it to 3 the shape would be (3,3) but we need (3,4)

```python
import numpy as np
a=np.arange(12).reshape(3,4)
b=np.arange(3).reshape(3)
print(a)
print(b)
print(a+b)
```

#### These two arrays can be broadcasted because we can add 1 to the head of arr2 and then stretch it to 2 and it will become of order (2,3)

```python
arr1=np.zeros((2,3))
arr2=np.ones((3))
print(arr1+arr2)
```

#### These two arrays can also be broadcasted because we can stretch 1 to 3 in both of arrays

```python
a = np.arange(3).reshape(1,3)
b = np.arange(3).reshape(3,1)

print(a)
print(b)

print(a+b)
```

#### These two are also following broadcasting rules , so they can also be broadcasted

```python
a = np.arange(3).reshape(1,3)
b = np.arange(4).reshape(4,1)

print(a)
print(b)

print(a + b)
```
