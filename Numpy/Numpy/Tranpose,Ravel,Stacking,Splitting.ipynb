Transpose of the Matrix

```python
import numpy as np
arr=np.arange(25).reshape(5,5)
```

```python
#Transpose-->Converts rows into columns and columns into rows
print(arr)
print("After Transpose")
print(arr.T)
```

```python
#Ravel converts any nd array to 1d array
arr2=np.array([[23,34],[1,45]])
arr2.ravel()
```

Stakcking of Matrices

```python
arr3=np.arange(10).reshape(2,5)
arr4=np.arange(10,20).reshape(2,5)
print(arr3)
print(arr4)
```

```python
#hstack stacks the matrices horizontally
np.hstack((arr3,arr4))
```

```python
#vstack stacks the matrices vertically
np.vstack((arr3,arr4))
```

Splitting (the opposite of stacking)

```python
#vsplit splits the arrays vertically
np.vsplit(arr3,2)
```

```python
#hsplit splits the arrays horizontally
arr5=np.arange(24).reshape(4,6)
np.hsplit(arr5,2)
```
