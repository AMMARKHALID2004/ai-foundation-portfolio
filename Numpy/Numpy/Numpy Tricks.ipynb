### np.clip

numpy.clip() function is used to Clip (limit) the values in an array.

https://numpy.org/doc/stable/reference/generated/numpy.clip.html

```python
#Values larger than a_max will become a_max and smaller than a_min will become a_min
array1=np.array([1,35,758,322,367,1352,2,56,23,6,99,24,11])
np.clip(array1,a_min=10,a_max=100)
```

### np.swapaxis

Interchange two axes of an array.This function is much more specific than transpose.It takes exactly two axes and swaps them, leaving all other axes exactly where they are.
For 2D arrays transpose and the swapaxis function do same thing
https://numpy.org/devdocs/reference/generated/numpy.swapaxes.html

```python
x = np.array([[1,2,3]])
np.swapaxes(x,0,1)
```

```python
x = np.array([[[0,1],[2,3]],[[4,5],[6,7]]])
x
np.swapaxes(x,0,2)
```
