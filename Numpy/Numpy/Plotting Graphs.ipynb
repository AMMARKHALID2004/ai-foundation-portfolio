# Sin Function

```python
import numpy as np
import matplotlib.pyplot as plt
x=np.linspace(-10,10,100)
y=(np.sin(x))
plt.plot(x,y)
```

# Straight line through origin

```python
x=np.linspace(-100,100,100)
y=x
plt.plot(x,y)
```

# Parabola

```python

x=np.linspace(-1000,1000,100)
y=x**2
plt.plot(x,y)
```

# Sigmoid Function

```python
x=np.linspace(-10,10,100)
y=1/(1+np.exp(-x))
plt.plot(x,y)
```
