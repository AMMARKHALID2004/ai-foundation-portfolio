```python
import numpy as np

arr=np.random.randint(1,200,24).reshape(6,4)

print(arr)
```

```python
#you can apply relational operators on whole matrix and the you can use the result of that
#to mask it on the original matrix and fetch the numbers of your choice
arr[arr>50]
```

```python
arr[(arr>50) & (arr%2==0)]
```

```python
arr[arr%7!=0]
```
