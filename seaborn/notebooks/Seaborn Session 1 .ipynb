```python
import seaborn as sns
import matplotlib.pyplot as plt
import plotly.express as px
```

### 1. Relational Plot

#### Scatter plot

```python
tips=sns.load_dataset('tips')
tips
```

```python
# axis level function
sns.scatterplot(data=tips,x='total_bill',y='tip')
```

```python
#figure level function (Figure level graphs are squarish by default) 
sns.relplot(data=tips,x='total_bill',y='tip',kind='scatter')
```

```python
#hue and style parameters can be used to get extra information
sns.relplot(data=tips,x='total_bill',y='tip',kind='scatter',hue='sex',size='size')
```

```python
sns.relplot(data=tips,x='total_bill',y='tip',kind='scatter',style='time',hue='sex')
```

#### Lineplot

```python
gap=px.data.gapminder()
gap
```

```python
df=gap[gap['country']=='Pakistan']
df
```

```python
#axis level function
sns.lineplot(data=df,x='year',y='lifeExp')
```

```python
#figure level function
sns.relplot(data=df,x='year',y='lifeExp',kind='line')
```

```python
temp_df=gap[gap['country'].isin(['Pakistan','United States','Australia','Afghanistan','Russia'])]
temp_df
```

```python
sns.relplot(kind='line',data=temp_df,x='year',y='lifeExp',hue='country',style='continent')
```

```python
#facetplot --> works with only figure level functions
sns.relplot(kind='line',data=temp_df,x='year',y='lifeExp',col='country',col_wrap=2)
```

```python
sns.relplot(kind='scatter',data=gap,y='gdpPercap',x='lifeExp',col='year',col_wrap=3,hue='country')
```

```python
sns.relplot(data=tips, x='total_bill', y='tip', kind='scatter', col='sex', row='day',hue='sex')
```

### 2. Distribution Plots

- used for univariate analysis
- used to find out the distribution
- Range of the observation
- Central Tendency
- is the data bimodal?
- Are there outliers?

Plots under distribution plot

- histplot
- kdeplot
- rugplot

#### Histplot

```python
#axis level
sns.histplot(data=tips,x='tip')
```

```python
#figure level
sns.displot(data=tips,x='tip')
```

```python
#element=step will change the appearance of the graph
sns.displot(data=tips,x='tip',element='step',bins=40)
```

```python
# It’s also possible to visualize the distribution of a categorical variable using the logic of a histogram. 
# Discrete bins are automatically set for categorical variables
sns.displot(data=tips,x='day')
```

```python
sns.displot(data=tips,x='total_bill',hue='sex',element='step')
```

```python
titanic=sns.load_dataset('titanic')
titanic
```

#### kdeplot
Rather than using discrete bins, a KDE plot smooths the observations with a Gaussian kernel, producing a continuous density estimate

```python
#figure level funtion
sns.displot(data=tips,x='tip',kind='kde',hue='sex')
```

```python
#fill=True will fill the graph
# height and aspect controls height and width respectively for figure level functions in seaborn as matplotlib's plt.figure(figsize=) will not
#work on the figure level functions
sns.displot(data=tips,x='tip',kind='kde',fill=True,height=10,aspect=2)
```

### Rugplot
Plots marginal distributions by drawing ticks along the x and y axes.
This function is intended to complement other plots by showing the location of individual observations in an unobtrusive way.

```python
#here we are using the axis level function because both graph will be plotten in the same axis, figure level graph cannot handle it
sns.kdeplot(data=titanic,x='fare')
sns.rugplot(data=titanic,x='fare')
```

### Bivariate Histogram

### Bivariate kdeplot

```python
sns.displot(data=tips,x='total_bill',y='tip',kind='kde')
```

### 3. Matrix plot
- Heatmap
- Clustermap

```python
temp=gap[gap['continent']=='Asia']
temp
```

```python
#heatmap is an axis level function only
temp=temp.pivot_table(index='country',columns='year',values='lifeExp')
temp=temp.drop(index='Israel')
temp
```

#### Heatmap

```python
#size of axis level functions can be increased using matplotlib
plt.figure(figsize=(15,10))
sns.heatmap(temp,cmap="viridis",annot=True,linewidths=1) #annot=True will show the valus and linewidth will make spaces in between the blocks
```

#### Clustermap
The clustermap function (most commonly used in the Python library Seaborn) creates a hierarchically-clustered heatmap. It visualizes a matrix dataset as a color-coded grid and automatically rearranges rows and columns to group similar data points together, displaying dendrograms along the axes

```python
iris=px.data.iris()
iris
```

```python
sns.clustermap(iris.iloc[:,0:4])
```
