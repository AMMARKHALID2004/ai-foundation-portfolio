```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.impute import KNNImputer
from sklearn.metrics import accuracy_score
from sklearn.linear_model import LogisticRegression
```

```python
df = pd.read_csv("/home/ammar/AI Track/Machine-Learning/Handling Missing Values/Multivariate Imputations/Datasets/train (2).csv",usecols = ['Age','Fare', 'Survived', 'Pclass'])
```

```python
df.head()
```

```python
X_train, X_test, y_train, y_test = train_test_split(df.iloc[:,1:], df['Survived'], test_size = 0.3, random_state=42)
```

```python
df.isna().sum()
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
s = StandardScaler()
X_train_trf = s.fit_transform(X_train)
X_test_trf = s.transform(X_test)
# Pclass is an ordinal categorical feature but for learning we can scale it too
```

```python
X_train_trf
```

```python
knn = KNNImputer(n_neighbors = 12)

X_train_imputed = knn.fit_transform(X_train_trf)
X_test_imputed = knn.transform(X_test_trf)
```

```python
pd.DataFrame(X_train_imputed, columns=X_train.columns, index=X_train.index)
```

```python
clf = LogisticRegression()
clf.fit(X_train_imputed, y_train)
```

```python
clf.predict(X_test_imputed)
```

```python
accuracy_score(y_test,clf.predict(X_test_imputed)) # accuracy will be changed by changing n_neighbors
```
