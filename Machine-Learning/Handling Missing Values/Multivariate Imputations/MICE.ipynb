### MICE --> Missing Imputation by Chained Equations
this works best on MAR  (missing at random) data (remember not MCAR missing completety at random data)

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import IterativeImputer
```

```python
df = pd.read_csv('/home/ammar/AI Track/Machine-Learning/Handling Missing Values/Multivariate Imputations/Datasets/train (2).csv', usecols=['Age','Survived','Fare'])
```

```python
df.head()
```

```python
X_train, X_test, y_train, y_test = train_test_split(df.iloc[:,1:],df['Survived'],test_size=0.3,random_state=42)
```

```python
df.isna().sum()
```

```python
iterative_i = IterativeImputer(max_iter=10, random_state=42)
```

```python
X_train_trf = iterative_i.fit_transform(X_train)
X_test_trf = iterative_i.transform(X_test)
```
