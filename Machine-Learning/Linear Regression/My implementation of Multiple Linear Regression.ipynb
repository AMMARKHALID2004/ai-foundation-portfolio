## Scikit Learn's Linear Regression Class

```python
import numpy as np
import pandas as pd
from sklearn.datasets import load_diabetes
```

```python
X,y = load_diabetes(return_X_y=True)
```

```python
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.3,random_state=42)
```

```python
lr = LinearRegression()
```

```python
y_pred = lr.predict(X_test)
y_pred
```

```python
from sklearn.metrics import r2_score
```

```python
r2_score(y_pred,y_test)
```

## Implementing my own class for Multiple Linear Regression
This will work for any number of inputs even one (for Simple Linear Regression)

```python
class MultipleLR():
    def __init__(self):
        self.coef_ = None   # this is beta (1 to n) or weights
        self.intercept_ = None # this is beta 0 or bias

    def fit(self,X_train,y_train):
        X_train = np.insert(X_train,0,1,axis=1) # inserting 1 at the start of every row because X matrix should have a column of 1s at the start
        calculations = np.linalg.inv(np.dot(X_train.T,X_train)).dot(X_train.T).dot(y_train)
        self.coef_ = calculations[1:]
        self.intercept_ = calculations[0]

    def pred(self,X_test):

        return np.dot(X_test,self.coef_) + self.intercept_
```

```python
myclass = MultipleLR()
```

```python
myclass.fit(X_train,y_train)
```

```python
y_pred = myclass.pred(X_test)
y_pred
```

```python
r2_score(myclass.pred(X_test),y_test) # r2 score is exactly the same
```
