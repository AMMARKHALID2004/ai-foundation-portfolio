```python
from sklearn.datasets import load_diabetes
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
X,y = load_diabetes(return_X_y=True)
```

```python
X.shape,y.shape
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X,y,random_state=42,test_size=0.2)
```

### My own class for Stochastic Gradient Descent Regressor

```python
class SGDRegressor:
    def __init__(self,learning_rate=0.01,epochs=100): #default values of learning rate and epochs given by me
        self.intercept_ = None
        self.coef_ = None
        self.lr = learning_rate
        self.epochs = epochs

    def fit(self,X_train,y_train):
        #random values of intercept and coefficients
        self.intercept_ = 0
        self.coef_ = np.ones(X_train.shape[1])

        for i in range(self.epochs):
            for j in range(X_train.shape[0]):
                idx = np.random.randint(0,X_train.shape[0])
                y_hat = np.dot(X_train[idx],self.coef_) + self.intercept_
    
                intercept_der = -2 * (y_train[idx] - y_hat)
                self.intercept_ = self.intercept_ - (self.lr * intercept_der)
                    
                coef_der = -2 * np.dot((y_train[idx] - y_hat),X_train[idx])
                self.coef_ = self.coef_ - (self.lr * coef_der)
        

    def predict(self,X_test):
        return np.dot(X_test,self.coef_) + self.intercept_
```

```python
sgd = SGDRegressor(learning_rate=0.01,epochs=400)
```

```python
import time
start = time.time()
sgd.fit(X_train,y_train)
print("The time taken is",time.time() - start,"which is much less than Batch Gradient Descent")
```

```python
from sklearn.metrics import r2_score
```

```python
y_pred = sgd.predict(X_test)
```

```python
r2_score(y_pred,y_test) # this answer will be different every time due to stochastic (random number choice) behaviour
```
