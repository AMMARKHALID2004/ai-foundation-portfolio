```python
from sklearn.datasets import load_diabetes
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
X,y = load_diabetes(return_X_y=True)
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(X,y,random_state=42,test_size=0.2)
```

### My own class for Gradient Descent Regressor

```python
class GDRegressor:
    def __init__(self,learning_rate=0.01,epochs=100): #default values of learning rate and epochs given by me
        self.intercept_ = None
        self.coef_ = None
        self.lr = learning_rate
        self.epochs = epochs

    def fit(self,X_train,y_train):
        #random values of intercept and coefficients
        self.intercept_ = 0
        self.coef_ = np.ones(X_train.shape[1])

        for i in range(self.epochs):
            
            y_hat = np.dot(X_train,self.coef_) + self.intercept_
            
            intercept_derivative = -2 * np.mean(y_train - y_hat)

            self.intercept_ = self.intercept_ - (self.lr * intercept_derivative)

            coef_der = -2 * (np.dot((y_train - y_hat),X_train)) / X_train.shape[0]
            self.coef_ = self.coef_ - (self.lr * coef_der) 

    def predict(self,X_test):
        return np.dot(X_test,self.coef_) + self.intercept_
```

```python
gdr = GDRegressor(learning_rate=0.1,epochs=1000)
```

```python
gdr.fit(X_train,y_train)
```

```python
gdr.predict(X_test)
```

```python
from sklearn.metrics import r2_score
```

```python
r2_score(gdr.predict(X_test),y_test)
```
