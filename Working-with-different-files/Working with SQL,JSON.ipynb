```python
from sqlalchemy import create_engine
```

```python
from pathlib import Path
import pandas as pd


# 1. This converts '~' into something like '/Users/yourname' or '/home/yourname'
db_path = Path().home()/'Machine Learning/Data Gathering/rideshare.db'

# 2. Build the engine using 4 slashes for the absolute path
conn = create_engine(f'sqlite:////{db_path}')

# 3. Your query
query = "SELECT * FROM drivers"

# 4. Load into Pandas
try:
    df = pd.read_sql_query(query, con=conn)
    print("Data loaded perfectly!")
except Exception as e:
    print(f"Error: {e}")
finally:
    conn.dispose()
```
