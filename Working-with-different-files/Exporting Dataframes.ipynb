#### Exporting as csv

```python
import pandas as pd
```

```python
df=pd.read_csv('~/Machine Learning/Pandas/Datasets/deliveries.csv')
df
```

```python
temp_df=df.groupby('batsman')['batsman_runs'].sum().reset_index()
temp_df.to_csv('Batsman Runs.csv',index=False) # index =False will stop the index from creating
```

#### Exporting as excel

```python
temp_df_2=df[df['batting_team']=='Mumbai Indians']
temp_df_2
```

```python
with pd.ExcelWriter('test.xlsx') as writer:
    temp_df.to_excel(writer,sheet_name='1')
    temp_df_2.to_excel(writer,sheet_name='2')
```

#### Exporting as HTML

```python
html=df.pivot_table(index='batsman',columns='bowling_team',values='batsman_runs',aggfunc='sum')
html
```

```python
html.to_html('batter_heatmap.html')
```

#### Exporting as JSON

```python
df.groupby(['batting_team','batsman'])['batsman_runs'].sum().unstack().to_json('ipl.json')
```

#### Exporting as a table of a database

```python
import pymysql
from sqlalchemy import create_engine

engine = create_engine("mysql+pymysql://root:@localhost/ipl")
# {root}:{password}@{url}/{database}
df.to_sql('ipl_delivery', con = engine, if_exists = 'append')

temp_df.to_sql('batsman_runs', con = engine, if_exists = 'append')

```
